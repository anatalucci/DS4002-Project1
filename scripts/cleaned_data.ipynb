# Clean Data

```python
import pandas as pd
df = pd.read_csv("../data/most_reviewed_asin.csv") # upload dataset
df.head()
```

```python
df = df[df["verified_purchase"] == True] # remove rows where verified purchase is false 
```

```python
len(df) 
df["verified_purchase"].value_counts() # check to verify rows were remvoed
```

```python
df["asin"].value_counts() # check to make sure all of the rows in the dataset are for the firestick 
```

```python
df.columns
```

```python
df = df.drop(columns=["images", "parent_asin", "user_id", "helpful_vote"]) # drop unnecessary columns
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce') # edit timestamp to only display year 
df['timestamp'] = df['timestamp'].dt.year.astype(str)
```

```python
df.to_csv("../data/firestick_reviews_cleaned.csv", index=False) # save to new csv
```
